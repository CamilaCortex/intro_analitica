## Pregunta 1. ¿Con qué Python estoy trabajando?

Python puede estar instalado varias veces en un mismo computador. Esta celda nos dice **cuál** está ejecutando este notebook, en qué **versión** y en qué **sistema operativo**.

```python
import platform
import sys

print("Sistema operativo :", platform.system())
print("Versión de Python :", platform.python_version())
print("Ubicación         :", sys.executable)
```

**¿Cómo leer el resultado?**

- La **versión** debe empezar por `3.14`.
- La **ubicación** debe incluir la carpeta `.venv` del proyecto. Según tu sistema, se verá parecido a esto:

| Sistema | Ejemplo de ubicación |
|---|---|
| macOS (`Darwin`) | `/Users/ana/intro_analitica/.venv/bin/python` |
| Linux | `/home/ana/intro_analitica/.venv/bin/python` |
| Windows | `C:\Users\Ana\intro_analitica\.venv\Scripts\python.exe` |

No importa que las rutas sean distintas: lo importante es que aparezca **`.venv`**.

## Pregunta 2. ¿Estoy dentro de la caja del proyecto?

Esta celda lo comprueba automáticamente y te dice "Sí" o "No".

```python
dentro_del_entorno = sys.prefix != sys.base_prefix

if dentro_del_entorno:
    print("Sí. Estás usando el entorno virtual del proyecto:")
    print("   ", sys.prefix)
else:
    print("No. Estás usando el Python general del computador.")
    print("Cambia el kernel a .venv (sección 8 del README) y vuelve a ejecutar la celda.")
```

**¿Cómo lo sabe?** Python guarda dos direcciones:

- `sys.base_prefix`: dónde está instalado el Python "original".
- `sys.prefix`: dónde está el Python que se está usando ahora.

Si son **distintas**, significa que estamos dentro de una caja (un entorno virtual) creada a partir de ese Python original.

## Pregunta 3. ¿Qué librerías tengo instaladas?

Las librerías del curso (pandas, scikit-learn, etc.) se instalan dentro de `.venv` cuando ejecutas `uv sync`. Esta celda revisa cada una y muestra su versión.

```python
from importlib import metadata

librerias = ["pandas", "numpy", "matplotlib", "seaborn", "scikit-learn", "fastapi"]

for libreria in librerias:
    try:
        version = metadata.version(libreria)
        print(f"{libreria:<14} instalada, versión {version}")
    except metadata.PackageNotFoundError:
        print(f"{libreria:<14} NO está instalada. Ejecuta `uv sync` en la terminal.")
```

Si alguna aparece como **no instalada**, abre una terminal en la carpeta del repositorio, ejecuta `uv sync` y vuelve a correr la celda.

## Pregunta 4. ¿Por qué todos debemos tener las mismas versiones?

Las librerías cambian con el tiempo, y el **mismo código** puede dar resultados distintos según la versión instalada.

**Ejemplo real:** a partir de pandas 3.0, las columnas de texto se guardan con un tipo de dato llamado `str`. En versiones anteriores ese tipo se llamaba `object`.

```python
import pandas as pd

ciudades = pd.DataFrame({"ciudad": ["Medellín", "Bogotá", "Cali"]})

print("Versión de pandas      :", pd.__version__)
print("Tipo de la columna     :", ciudades["ciudad"].dtype)
```

Con pandas 3 verás `str`; un compañero con pandas 2 vería `object`. Diferencias como esta pueden hacer que un código funcione en un computador y falle en otro.

Por eso el proyecto tiene el archivo **`uv.lock`**: guarda la versión exacta de cada librería, y cuando cualquier persona ejecuta `uv sync` obtiene **exactamente las mismas**.
